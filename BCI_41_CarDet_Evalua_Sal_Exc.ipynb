### Evaluacion: Excepcion operaciones con monto deuda ifrs cero
---
* Operaciones con deuda total ifrs cero, pero es del tipo Nova CCN455, COLMN CON731 Y COLMN COM732
* flag_resultado_regla =1: significa que es una excepcion a la regla de salida por saldo total ifrs cero

```python
paso_query30 =  f"""
CREATE OR REPLACE TEMPORARY VIEW tmp_RES_D00_OPE_CAMPO_EVAL_1 as
SELECT
    A.periodo_cierre                                            AS periodo_cierre,
    A.fecha_cierre                                              AS fecha_cierre,
    A.tipo_proceso                                              AS tipo_proceso,
    A.segmento                                                  AS segmento,
    A.operacion                                                 AS operacion,
    A.tipo_operacion                                            AS tipo_operacion,
    A.sistema                                                   AS sistema,
    A.rut_cliente                                               AS rut_cliente,
    A.dv_rut_cliente                                            AS dv_rut_cliente ,
    'saldo_total_ifrs - (sistema - tipo_operacion)'               AS nombre_campo,
    concat('[',cast(trim(A.saldo_total_ifrs) as string),'-',cast(trim(A.sistema) as string),'-',cast(trim(A.tipo_operacion) as string),']')        AS valor_campo,
    'Excepcion saldo capital ifrs cero para'   AS condicion_regla,     
    '[eq to 0 - AND  (20-CCN455, 01-CON731, 01-COM732)]'                AS valor_regla,           
    CASE 
       WHEN trim(A.sistema) = '20' AND trim(A.tipo_operacion)='CCN455' THEN 1
       WHEN trim(A.sistema) = '01' AND trim(A.tipo_operacion)='CON731' THEN 1
       WHEN trim(A.sistema) = '01' AND trim(A.tipo_operacion)='COM732' THEN 1
       ELSE 0
    END                                                          AS flag_resultado_regla,
    'N01'                                                        AS cod_evaluacion  
FROM 
    {base_silver_x}.tbl_cd_ope_condicion_salida_deterioro A
WHERE 
    IFNULL(A.saldo_total_ifrs,0) = {p_crit_det_sal_tot_ifrs}
"""
```

```python
sql_safe(paso_query30)
```

### Evaluacion: Excepcion operaciones con monto deuda ifrs cero, PERO CLIENTE LIR
---
* Operaciones con deuda total ifrs cero, pero es cliente lir 
* flag_resultado_regla =1: significa que es una excepcion a la regla de salida por saldo total ifrs cero


```python
paso_query40 =  f"""
CREATE OR REPLACE TEMPORARY VIEW tmp_RES_D00_OPE_CAMPO_EVAL_2 as
SELECT
    A.periodo_cierre                                            AS periodo_cierre,
    A.fecha_cierre                                              AS fecha_cierre,
    A.tipo_proceso                                              AS tipo_proceso,
    A.segmento                                                  AS segmento,
    A.operacion                                                 AS operacion,
    A.tipo_operacion                                            AS tipo_operacion,
    A.sistema                                                   AS sistema,
    A.rut_cliente                                               AS rut_cliente,
    A.dv_rut_cliente                                            AS dv_rut_cliente ,
    'saldo_total_ifrs - flag_existe_cliente_lir - cli_fecha_informada_lir'               AS nombre_campo,
    concat('[',cast(trim(A.saldo_total_ifrs) as string),'-', cast(IFNULL(A.flag_existe_cliente_lir,0) as string),'-',cast(trim(A.cli_fecha_informada_lir) as string),']')      AS valor_campo,
    'Excepcion saldo capital ifrs cero para cliente LIR'           AS condicion_regla,     
    "[0 - 1 - <= {p_fec_12_meses_atras}]"                AS valor_regla,        
    CASE 
       WHEN IFNULL(A.flag_existe_cliente_lir,0) = 1 AND A.cli_fecha_informada_lir>{p_fec_12_meses_atras} THEN 1
       ELSE 0
    END                                                          AS flag_resultado_regla,
    'N02'                                                        AS cod_evaluacion  
FROM 
    {base_silver_x}.tbl_cd_ope_condicion_salida_deterioro  A
WHERE 
    IFNULL(A.saldo_total_ifrs,0) = {p_crit_det_sal_tot_ifrs}
"""
```

```python
sql_safe(paso_query40)
```

### Evaluacion: Excepcion operaciones con monto deuda ifrs cero, PERO operacion INDIVIDUAL
---
* Operaciones con deuda total ifrs cero, pero operacion  es segmento individual
* flag_resultado_regla =1: significa que es una excepcion a la regla de salida por saldo total ifrs cero


```python
paso_query50 =  f"""
CREATE OR REPLACE TEMPORARY VIEW tmp_RES_D00_OPE_CAMPO_EVAL_3 as
SELECT
    A.periodo_cierre                                            AS periodo_cierre,
    A.fecha_cierre                                              AS fecha_cierre,
    A.tipo_proceso                                              AS tipo_proceso,
    A.segmento                                                  AS segmento,
    A.operacion                                                 AS operacion,
    A.tipo_operacion                                            AS tipo_operacion,
    A.sistema                                                   AS sistema,
    A.rut_cliente                                               AS rut_cliente,
    A.dv_rut_cliente                                            AS dv_rut_cliente ,
    'saldo_total_ifrs - segmento'               AS nombre_campo,
    concat('[',cast(trim(A.saldo_total_ifrs) as string),'-',cast(A.segmento as string),']')      AS valor_campo,
    'Excepcion saldo capital ifrs cero para cliente Individual'           AS condicion_regla,     
    '[0-I] '                                    AS valor_regla,           
    CASE 
       WHEN trim(A.segmento) = 'I' THEN 1
       ELSE 0
    END                                                          AS flag_resultado_regla,
    'N03'                                                        AS cod_evaluacion  
FROM 
    {base_silver_x}.tbl_cd_ope_condicion_salida_deterioro A
WHERE 
    IFNULL(A.saldo_total_ifrs,0) = {p_crit_det_sal_tot_ifrs}
"""
```

```python
sql_safe(paso_query50)
```

### Evaluacion: Excepcion operaciones con monto deuda ifrs cero, PERO operacion Filiales
---
* Operaciones con deuda total ifrs cero, pero operacion  Filiales
* flag_resultado_regla =1: significa que es una excepcion a la regla de salida por saldo total ifrs cero


```python
paso_query60 =  f"""
CREATE OR REPLACE TEMPORARY VIEW tmp_RES_D00_OPE_CAMPO_EVAL_4 as
SELECT
    A.periodo_cierre                                            AS periodo_cierre,
    A.fecha_cierre                                              AS fecha_cierre,
    A.tipo_proceso                                              AS tipo_proceso,
    A.segmento                                                  AS segmento,
    A.operacion                                                 AS operacion,
    A.tipo_operacion                                            AS tipo_operacion,
    A.sistema                                                   AS sistema,
    A.rut_cliente                                               AS rut_cliente,
    A.dv_rut_cliente                                            AS dv_rut_cliente ,
    'saldo_total_ifrs - flag_existe_cliente_ssff - flag_existe_cliente_fact'               AS nombre_campo,
    concat('[',cast(trim(A.saldo_total_ifrs) as string),'-',cast(A.flag_existe_cliente_ssff as string),'-',cast(A.flag_existe_cliente_fact as string),']')      AS valor_campo,
    'Excepcion saldo capital ifrs cero para cliente de filiales'           AS condicion_regla,     
    "[0-(1 OR 1)] "                                    AS valor_regla,           
    CASE 
       WHEN IFNULL(A.flag_existe_cliente_ssff,0) = 1 OR IFNULL(A.flag_existe_cliente_fact,0) = 1 THEN 1
       ELSE 0
    END                                                          AS flag_resultado_regla,
    'N04'                                                        AS cod_evaluacion  
FROM 
    {base_silver_x}.tbl_cd_ope_condicion_salida_deterioro A
WHERE 
    IFNULL(A.saldo_total_ifrs,0) = {p_crit_det_sal_tot_ifrs}
"""
```

```python
sql_safe(paso_query60)
```

### Evaluacion: Excepcion operaciones grupal que sale de deterioro por Irradiación Intersegmento.
---
* Operaciones Grupal con deterioro, pero que tambien contiene el cliente operaciones  de segmento individual sin deterioro.
* flag_resultado_regla =1: significa que es una excepcion a la regla de salida 

```python
paso_query70 =  f"""
CREATE OR REPLACE TEMPORARY VIEW tmp_RES_D00_OPE_CAMPO_EVAL_5 as
SELECT
    A.periodo_cierre                                            AS periodo_cierre,
    A.fecha_cierre                                              AS fecha_cierre,
    A.tipo_proceso                                              AS tipo_proceso,
    A.segmento                                                  AS segmento,
    A.operacion                                                 AS operacion,
    A.tipo_operacion                                            AS tipo_operacion,
    A.sistema                                                   AS sistema,
    A.rut_cliente                                               AS rut_cliente,
    A.dv_rut_cliente                                            AS dv_rut_cliente ,
    'cli_segmento_cliente - segmento - cli_calificacion_bci'               AS nombre_campo,
    concat('[',cast(A.cli_segmento_cliente as string),'-',cast(A.segmento as string),'-',cast(A.cli_calificacion_bci as string),']')      AS valor_campo,
    "Cliente individual no deteriorado con operaciones grupales deterioradas (intersegmento)"           AS condicion_regla,     
    "[I - G - {param_cal}]"                           AS valor_regla,           
    CASE 
       WHEN trim(IFNULL(A.cli_calificacion_bci,'0')) NOT IN {param_cal} THEN 1
       ELSE 0
    END                                                          AS flag_resultado_regla,
    'N05'                                                        AS cod_evaluacion  
FROM 
   {base_silver_x}.tbl_cd_ope_condicion_salida_deterioro A
WHERE
    A.cli_segmento_cliente='I' and A.segmento='G'
"""
```

```python
sql_safe(paso_query70)
```
