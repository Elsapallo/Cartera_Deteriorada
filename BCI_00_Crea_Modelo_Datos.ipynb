### tbl_cd_cartdet_crit_ent_ope_eval (resultado de evaluaciones de entrada a deterioro)

 - Almacena todas las evaluaciones aplicadas a las operaciones, dejando un indicador si cumple o no cumple dicha regla.
 - Una operación puede tener muchas evaluaciones. Evaluciones por mora, renegociado, individual, factoring, ssff, etc.
 - En esta tabla no se asigna criterios de deterioros
 - Entonces la llave de esta tabla debe ser la operación y el codigo de evaluacion

```python
%sql			
CREATE TABLE IF NOT EXISTS ${bci.dbnamesilver}.tbl_cd_cartdet_crit_ent_ope_eval(			
 periodo_cierre	int	COMMENT	'Año y mes del campo fecha_cierre, el cual identifica de cuando es la información cargada'
,fecha_cierre	int	COMMENT	'año, mes y dia que componen la fecha correspondiente a la información cargada'
,tipo_proceso	string	COMMENT	'El tipo de proceso contiene la sigla C (Cierre) o PC (Precierre), que identifica de cuando es extraída la información cargada en esta tabla'
,segmento	string	COMMENT	'Código de segmento el cual se obtiene directo de tabla de salida del proceso Segmentacion, donde se expresan valores como por ejemplo (I, G)'
,operacion	string	COMMENT	'Corresponde al número de la operacion relacionada al cliente, esta puede estar compuesta de letras y numeros, como por ejemplo (A01020073786,C01361425478,D01010589707,E00000463489)'
,tipo_operacion	string	COMMENT	'Identifica al tipo y sub-tipo (Tio-Aux) de operación relacionado al cliente, esta es una sigla compuesta de 6 caracteres, como por ejemplo (CON447, 110459, COM604)'
,sistema	string	COMMENT	'Corresponde a un código que identifica de que sistema viene la operación relacionada al cliente, este código tiene dos dígitos, com por ejemplo (01, 02, 03, 04, 05)'
,rut_cliente	int	COMMENT	'Rut de cliente deteriorado'
,dv_rut_cliente	string	COMMENT	'Digito verificador cliente deteriorado'
,nombre_campo	string	COMMENT	'Código que identifica el criterio evaluado. Datos posibles del campo (MOT_BANCA_IND, MOT_DEUDA_TOT, MOT_DEUDA_PROM, MOT_GRUPO_EMP)'
,valor_campo	string	COMMENT	'Valor del criterio evaluado. Datos posibles del campo (G,C6, I,B4)'
,condicion_regla	string	COMMENT	'Condición del criterio evaluado. Datos posibles del campo (like, IN, >)'
,valor_regla	string	COMMENT	'Valor de la regla. Datos posibles del campo (I, C%, I2, I3, I4, 20000, tbl_grupos_empresariales)'
,flag_resultado_regla	tinyint	COMMENT	'Flag resultado (0 ó 1), el cual va a depender de la regla indicada por el negocio, para cada criterio. Cuando cumple es 1, de lo contrario es 0'
,cod_evaluacion	string	COMMENT	'codigo evaluacion'
)			
USING DELTA			
PARTITIONED BY (fecha_cierre)			
COMMENT 'Tabla que contiene las operaciones con su criterio evaluado de entrada a cartera deteriorada'			
LOCATION '${bci.ruta_silver}/tbl_cd_cartdet_crit_ent_ope_eval';			
```

 - Una operación puede ser deteriorada por varios criterios y aquí debe estar todos los criterios con que se evaluaron.

```python
 %sql 			
CREATE TABLE IF NOT EXISTS ${bci.dbnamesilver}.tbl_cd_cartdet_crit_ent_crit(			
 periodo_cierre	int	COMMENT	'Año y mes del campo fecha_cierre, el cual identifica de cuando es la información cargada'
,fecha_cierre	int	COMMENT	'año, mes y dia que componen la fecha correspondiente a la información cargada'
,tipo_proceso	string	COMMENT	'El tipo de proceso contiene la sigla C (Cierre) o PC (Precierre), que identifica de cuando es extraída la información cargada en esta tabla'
,rut_cliente	int	COMMENT	'Rut de cliente deteriorado'
,dv_rut_cliente	string	COMMENT	'Digito verificador cliente deteriorado'
,tipo_operacion	string	COMMENT	'Identifica al tipo y sub-tipo (Tio-Aux) de operación relacionado al cliente, esta es una sigla compuesta de 6 caracteres, como por ejemplo (CON447, 110459, COM604)'
,operacion	string	COMMENT	'Corresponde al número de la operacion relacionada al cliente, esta puede estar compuesta de letras y numeros, como por ejemplo (A01020073786,C01361425478,D01010589707,E00000463489)'
,sistema	string	COMMENT	'Corresponde a un código que identifica de que sistema viene la operación relacionada al cliente, este código tiene dos dígitos, com por ejemplo (01, 02, 03, 04, 05)'
,segmento	string	COMMENT	'Código de segmento el cual se obtiene directo de tabla de salida del proceso Segmentacion, donde se expresan valores como por ejemplo (I, G)'
,criterio_entrada	int	COMMENT	'Criterio de entrada a deterioro por operación'
,origen_deterioro	int	COMMENT	'Origen de deterioro'
,fecha_entrada	int	COMMENT	'Se calcula con el campo fecha_inicio_mora , cuando el dato es null o 0 se deja 19000101, de lo contrario se deja fecha_inicio_mora que es la fecha conformada de año, mes y dia en la que el crédito cae en mora'
,grupo	string	COMMENT	'Inidca la regla de de negocio por el cual la operacion se debe evaluar, esta se determina de la siguiente forma (SSFF, Factoring, BCI_Grupal, Bci_Irradiacion, BCI_Individual)'
,periodo_evaluacion          string		COMMENT	'Numenclatura que indica cuando la operacion fue evaluada. Los datos que se encuentran en este campo es p_anterior o p_actual)'
)			
USING DELTA			
PARTITIONED BY (fecha_cierre)			
COMMENT 'Tabla que contiene las operaciones que entran a cartera deteriorada con su criterio de entrada '			
LOCATION '${bci.ruta_silver}/tbl_cd_cartdet_crit_ent_crit';			
```
